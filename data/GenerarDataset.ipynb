```python
import pandas as pd
# recargar el CSV guardado
dfY = pd.read_csv("INDECOPI_OPS1.csv", sep=";")
```

**GENERAR VARIABLE - VC_LUGAR**

```python
# Crear nuevo dataset dfX
n = 1000000

# Generar dataset final dfX
# Primera columna - unico valor - SEDE CENTRAL
dfX = pd.DataFrame({
    "VC_LUGAR": ["SEDE CENTRAL"] * n
})

dfX.info()
```

**GENERAR VARIABLE - VC_TIPO_EXPEDIENTE**
- Se extrae la probabilidad de aparición de cada tipo de expediente en base al dataset original.
- Se generan datos mantenimiento el ratio original.

```python
# Para seleccionar los tipos de expediente / sacar probabilidad-ratio de aparicion

dist_tipoExp = dfY["VC_TIPO_EXPEDIENTE"].value_counts(normalize=True)
valores_tipoExp = dist_tipoExp.index
probs_tipoExp = dist_tipoExp.values

print("Tipos de expediente: ", valores_tipoExp)
print("Probabilidades: ", probs_tipoExp)
```

```python
# Generar datos de la columna VC_TIPO_EXPEDIENTE
import numpy as np

dfX["VC_TIPO_EXPEDIENTE"] = np.random.choice(
    valores_tipoExp,
    size=len(dfX),
    p=probs_tipoExp
)

dfX.info()
dfX["VC_TIPO_EXPEDIENTE"].unique()
```

```python
# Revisar resultados
dist_tipoExp = dfX["VC_TIPO_EXPEDIENTE"].value_counts(normalize=True)
print("Tipos de expediente: ", dist_tipoExp.index)
print("Probabilidades: ", dist_tipoExp.values)
```

Mantiene una distribución similar a la original.

**GENERAR VARIABLE - EMPLEADO**

```python
# Realizar mismo proceso con EMPLEADO

dist_empl = dfY["EMPLEADO"].value_counts(normalize=True)
valores_empl = dist_empl.index
probs_empl = dist_empl.values

dfX["EMPLEADO"] = np.random.choice(
    valores_empl,
    size=len(dfX),
    p=probs_empl
)

dfX.info()
print(dfX["EMPLEADO"].unique().size)
dfX["EMPLEADO"].unique()
```

**GENERAR VARIABLE - FEC_PRE**
- Se genera fechas random en años posteriores a 2016.

```python
# Generar fechas random

# rango de fechas
inico = np.datetime64('2017-01-01')
fin = np.datetime64('2025-06-30')

# generar fechas aleatorias
fechas = np.random.randint(inico.astype('int64'), fin.astype('int64'), n)

# convertir a datetime
dfX["FEC_PRE"] = pd.to_datetime(fechas)

dfX.info()
```

**GENERAR VARIABLE - FEC_REING**
- Considerando que puede ser nulo. Se buscar mantener un ratio nulo-fecha similar al dataset original.
- La fecha generada debe ser posterior a FEC_PRE.

```python
# FEC_REING se calcula considerando una probabilidad similar a la original
prop_reing = dfY["FEC_REING"].notna().mean()

# Mantener distribucion de dias entre presentacion y reingreso
dfY_temp = dfY.dropna(subset=["FEC_PRE", "FEC_REING"]).copy()

dfY_temp["dias"] = (pd.to_datetime(dfY_temp["FEC_REING"]) - 
                    pd.to_datetime(dfY_temp["FEC_PRE"])).dt.days

dias_reing = dfY_temp["dias"]
dias_reing = dias_reing[dias_reing >= 0] # solo valores validos

# Decidir los registros no nulos
mascara_reing = np.random.rand(n) < prop_reing
# Elegir dias random
dias_muestra = np.random.choice(dias_reing, size=n)
# Sumar dias random
fec_reing = dfX["FEC_PRE"] + pd.to_timedelta(dias_muestra, unit="D")
# Limitar fecha
max_fecha = pd.to_datetime("2025-12-31")
fec_reing = fec_reing.where(fec_reing <= max_fecha, max_fecha)
# Crear la columna "FEC_REING" considerando los nulos
dfX["FEC_REING"] = pd.NaT
dfX.loc[mascara_reing, "FEC_REING"] = fec_reing[mascara_reing]

dfX.info()
```

**GENERAR VARIABLE - NRO_EXPEDIENTE**
- Es un código identificador que aumenta de 1 en 1.
- El orden depende de la fecha de presentación.

```python
# Generar numero de expediente NRO_EXPEDIENTE

# ordenar por fecha
dfX = dfX.sort_values("FEC_PRE").reset_index(drop=True)

# Extraer-armar datos
dfX["ANIO"] =dfX["FEC_PRE"].dt.year
dfX["NUMCORR"] = dfX.groupby("ANIO").cumcount()+1
dfX["NUMCORR"] = dfX["NUMCORR"].astype(str).str.zfill(6) # llenar con 0

# Construir NRO_EXPEDIENTE
dfX["NRO_EXPEDIENTE"] = (dfX["NUMCORR"] + "-" + dfX["ANIO"].astype(str) + "/PS1")

# Eliminar columnas temporales
dfX = dfX.drop(columns=["ANIO","NUMCORR"])

dfX.info()
dfX.head(4)
```

**GENERAR VARIABLE - FEC_CAL**
- No es nulo.
- Sigue un proceso similar a FEC_REING. Es posterior a FEC_PRE.

**GENERAR VARIABLE - DESCR_CIIU**
- Depende de la variable VC_RUBRO.
- Un VC_RUBRO puede tener multiples DESCR_CIIU.
- Se mantiene una distribución en DESCR_CIIU en base a VC_RUBRO similar ala original.

```python
from collections import defaultdict

# DESCR_CIIU es un campo dependiente de VC_RUBRO.

df_temp3 = dfY.dropna(subset=["VC_RUBRO", "DESCR_CIIU"]).copy()
dist_condicional = (
    df_temp3
    .groupby("VC_RUBRO")["DESCR_CIIU"]
    .value_counts(normalize=True)
) # Existen probabilidade distitnas de DESCR_CIIU para cada VC_RUBRO

# Usar diccionario para facilidad de acceso
dic_dist = defaultdict(lambda: ([], []))

for (rubro, ciiu), prob in dist_condicional.items():
    dic_dist[rubro][0].append(ciiu) # valor CIIU
    dic_dist[rubro][1].append(prob) # probabilida del valor

def obtener_ciiu(rubro):
    if rubro in dic_dist:
        valores, probs = dic_dist[rubro]
        return np.random.choice(valores, p=probs)
    else:
        return np.nan   # no existe el rubro
    
# Generar el campo
dfX["DESCR_CIIU"] = dfX["VC_RUBRO"].apply(obtener_ciiu)
```
