```python
import pandas as pd
import numpy as np

path = '/content/INDECOPI_OPS1_Millon.csv'
df = pd.read_csv(path, sep=None, engine='python')
```

```python
stats_df = pd.DataFrame({
    'Type': df.dtypes,
    'Unique Values': df.nunique(),
    'Missing Values': df.isnull().sum(),
    'Missing %': (df.isnull().sum() / len(df) * 100).round(2)
})

print('Summary Statistics per Column:')
display(stats_df)

def get_duplicate_columns(df):
    duplicate_column_names = set()
    for x in range(df.shape[1]):
        col = df.iloc[:, x]
        for y in range(x + 1, df.shape[1]):
            other_col = df.iloc[:, y]
            if col.equals(other_col):
                duplicate_column_names.add(df.columns.values[y])
    return list(duplicate_column_names)

duplicates = get_duplicate_columns(df)
print(f'\nRepeated (identical) columns found: {duplicates if duplicates else "None"}')

print(f'\nShape of the dataset: {df.shape}')
display(df.head())
```

```python
cols_to_keep = [col for col in df.columns if df[col].nunique() > 2]
cols_to_drop = [col for col in df.columns if df[col].nunique() <= 2]

df_filtered = df[cols_to_keep].copy()

print(f'Columns removed: {cols_to_drop}')
print(f'Original shape: {df.shape}')
print(f'New shape: {df_filtered.shape}')
display(df_filtered.head())
```

```python
df_filtered.head()
```

```python
null_pct = df_filtered.isnull().sum() / len(df_filtered) * 100

cols_with_many_nulls = null_pct[null_pct > 10].index.tolist()

cols_to_drop_manually = ['EMPLEADO', 'NRO_EXPEDIENTE']
df_filtered = df_filtered.drop(columns=cols_with_many_nulls + cols_to_drop_manually)

print(f'Columns removed (>10% nulls): {cols_with_many_nulls}')
print(f'Columns removed manually: {cols_to_drop_manually}')
print(f'Remaining columns: {df_filtered.columns.tolist()}')
print(f'Final shape: {df_filtered.shape}')
display(df_filtered.head())
```

```python
final_stats = pd.DataFrame({
    'Type': df_filtered.dtypes,
    'Unique Values': df_filtered.nunique(),
    'Missing Values': df_filtered.isnull().sum(),
    'Missing %': (df_filtered.isnull().sum() / len(df_filtered) * 100).round(2)
})

print('Final Dataset Summary Statistics:')
display(final_stats)
```

```python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text) or not isinstance(text, str):
        return text
    text = text.upper()
    text = ''.join(c for c in unicodedata.normalize('NFKD', text) if unicodedata.category(c) != 'Mn')
    text = re.sub(r'[^A-Z0-9 ]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

cols_to_normalize = ['VC_TIPO_EXPEDIENTE', 'DENUNCIANTES', 'DENUNCIADOS']

for col in cols_to_normalize:
    print(f'Normalizing {col}...')
    df_filtered[col] = df_filtered[col].apply(normalize_text)

print('\nNormalization complete. Preview of cleaned data:')
display(df_filtered.head())
```

```python
def determine_target_key(row):
    ruc = str(row['RUC_DENUNCIADOS']) if pd.notna(row['RUC_DENUNCIADOS']) else ''
    if len(ruc) == 11 and ruc.isdigit():
        return ruc
    return row['DENUNCIADOS']

df_filtered['USER_KEY'] = df_filtered['DENUNCIANTES']
df_filtered['TARGET_KEY'] = df_filtered.apply(determine_target_key, axis=1)
df_filtered['TYPE_KEY'] = df_filtered['VC_TIPO_EXPEDIENTE']

print('New key columns created:')
display(df_filtered[['USER_KEY', 'TARGET_KEY', 'TYPE_KEY']].head())
```
