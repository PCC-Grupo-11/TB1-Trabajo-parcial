```python
import os
import json
import random
from pathlib import Path
import kagglehub

dataset_path = kagglehub.dataset_download("yelp-dataset/yelp-dataset")
print("Path to dataset files:", dataset_path)

dataset_root = Path(dataset_path)
review_candidates = list(dataset_root.rglob("yelp_academic_dataset_review.json"))
if not review_candidates:
    raise FileNotFoundError("No se encontró yelp_academic_dataset_review.json en la ruta descargada.")

review_file = review_candidates[0]
print("Review file:", review_file)
```

```python
import polars as pl

SAMPLE_SIZE = 100_000
SEED = 10

df_sample = (
    pl.scan_ndjson(str(review_file))
    .collect()
    .sample(n=SAMPLE_SIZE, seed=SEED)
)

print(f"Muestra de {len(df_sample):,} filas creada con éxito.")
```

```python
output_path_csv = Path("/content/yelp_review_sample_100k.csv")

df_sample.write_csv(output_path_csv)

print("Archivo CSV de muestra generado con Polars:", output_path_csv)
print(f"Tamaño (MB): {output_path_csv.stat().st_size / (1024 * 1024):.2f}")

display(df_sample.head(3))
```
